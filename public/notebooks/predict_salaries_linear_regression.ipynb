## Step 1: Environment Setup & Data Ingestion
We import standard data science libraries (`pandas`, `numpy`, `matplotlib.pyplot`, `seaborn`, `sklearn`) and load the dataset from the public repository.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load the Salary Dataset
url = 'https://raw.githubusercontent.com/krishnaik06/simple-Linear-Regression/master/Salary_Data.csv'
df = pd.read_csv(url)

print('=== Dataset First 5 Rows ===')
print(df.head())

print(f'\n=== Dataset Shape: {df.shape} ===')

print('\n=== Missing Values Check ===')
print(df.isnull().sum())
```

## Step 2: Exploratory Data Analysis (EDA)
Let's inspect summary statistics and calculate the Pearson correlation coefficient between Experience and Salary.

```python
print('=== Summary Statistics ===')
print(df.describe())

corr = df['YearsExperience'].corr(df['Salary'])
print(f'\nPearson Correlation Coefficient: r = {corr:.4f}')

# Visualizing the relationship
plt.figure(figsize=(8, 5))
sns.scatterplot(x='YearsExperience', y='Salary', data=df, color='#001f54', s=70)
plt.title('Years of Experience vs. Salary ($)', fontsize=13, fontweight='bold', color='#001f54')
plt.xlabel('Years of Experience')
plt.ylabel('Salary in USD ($)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## Step 3: Feature Matrix & Train-Test Split
We extract $X$ (2D feature matrix) and $y$ (1D target vector), and partition into 2/3 training data (20 samples) and 1/3 test data (10 samples).

```python
X = df[['YearsExperience']]
y = df['Salary']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=1/3, random_state=0)

print(f'X_train shape: {X_train.shape}')
print(f'X_test shape:  {X_test.shape}')
print(f'y_train shape: {y_train.shape}')
print(f'y_test shape:  {y_test.shape}')
```

## Step 4: Training the Linear Regression Model
We fit `LinearRegression()` on `X_train` and inspect the learned parameters (Slope $\beta_1$ and Intercept $\beta_0$).

```python
regressor = LinearRegression()
regressor.fit(X_train, y_train)

slope = regressor.coef_[0]
intercept = regressor.intercept_

print('=== Model Parameters ===')
print(f'Slope (Weight beta_1):     ${slope:,.2f} per year of experience')
print(f'Intercept (Bias beta_0):   ${intercept:,.2f} starting baseline salary')
print(f'Learned Formula: Salary = {slope:.2f} * YearsExperience + {intercept:.2f}')
```

## Step 5: Model Evaluation & Metric Breakdown
We evaluate the model on the unseen test set using MAE, MSE, RMSE, and $R^2$.

```python
y_pred = regressor.predict(X_test)

test_mae = mean_absolute_error(y_test, y_pred)
test_mse = mean_squared_error(y_test, y_pred)
test_rmse = np.sqrt(test_mse)
test_r2 = r2_score(y_test, y_pred)

results_df = pd.DataFrame({
    'Experience': X_test['YearsExperience'].values,
    'Actual ($)': y_test.values,
    'Predicted ($)': np.round(y_pred, 2),
    'Residual Error ($)': np.round(y_test.values - y_pred, 2)
})

print('=== Test Set Predictions vs Actual ===')
print(results_df.to_string())

print('\n=== Performance Metrics ===')
print(f'Mean Absolute Error (MAE):       ${test_mae:,.2f}')
print(f'Mean Squared Error (MSE):        {test_mse:,.2f} ($^2)')
print(f'Root Mean Squared Error (RMSE):  ${test_rmse:,.2f}')
print(f'R-Squared Score (R²):            {test_r2:.4f} ({test_r2*100:.1f}% Variance Explained)')
```

## Step 6: Visualizing Training & Test Set Results
Let's visually plot the regression line against both the Training set and Test set.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Training Set Fit
axes[0].scatter(X_train, y_train, color='#0284c7', label='Train Data (N=20)')
axes[0].plot(X_train, regressor.predict(X_train), color='#001f54', linewidth=2.5, label='Regression Line')
axes[0].set_title('Training Set Fit', fontweight='bold', color='#001f54')
axes[0].set_xlabel('Years of Experience')
axes[0].set_ylabel('Salary ($)')
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.5)

# 2. Test Set Generalization
axes[1].scatter(X_test, y_test, color='#059669', label='Test Data (N=10)')
axes[1].plot(X_train, regressor.predict(X_train), color='#001f54', linewidth=2.5, label='Regression Line')
axes[1].set_title('Test Set Evaluation (R² = 0.975)', fontweight='bold', color='#001f54')
axes[1].set_xlabel('Years of Experience')
axes[1].set_ylabel('Salary ($)')
axes[1].legend()
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

## Step 7: Production Salary Prediction Function
We package our trained model into a production function.

```python
def predict_salary(years_of_experience):
    """Predicts employee salary based on years of experience."""
    exp_arr = np.array([[float(years_of_experience)]])
    predicted_salary = regressor.predict(exp_arr)[0]
    return predicted_salary

test_cases = [0.0, 3.5, 7.0, 12.0, 15.0]
for exp in test_cases:
    sal = predict_salary(exp)
    print(f'Predicted Salary for {exp:5.1f} yrs experience: ${sal:,.2f}')
```
